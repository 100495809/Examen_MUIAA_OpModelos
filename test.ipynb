```python
from typing import Literal
from pydantic import BaseModel, ConfigDict, Field, field_validator
from pathlib import Path
import csv
import joblib
```

```python
class Entrada(BaseModel):
    # TODO
    id_paquete: str = Field(min_length=1)

    peso_kg: float = Field(ge=0, le=30)

    distancia_km: float = Field()

    @field_validator("id_paquete",mode="after")
    @classmethod
    def delete_spaces(cls, id_paquete):
        try:
            id_paquete=id_paquete.strip()
        except ValueError as e:
            print(f"Error {e} al tratar la fila id_paquete")
        return id_paquete
```

```python
def leer_csv(ruta: Path) -> list[Entrada]:
    # TODO
    lista = []
    with open(ruta, mode='r', encoding='utf-8') as archivo:
        fichero = csv.DictReader(archivo)
    
        for fila in fichero:
            validacion = Entrada(id_paquete=fila['id_paquete'], peso_kg=fila['peso_kg'], distancia_km=fila['distancia_km'])
            validacion.model_dump()
            lista.append(validacion)
            # print(f"id_paquete: {fila['id_paquete']}, Peso: {fila['peso_kg']}, Distancia: {fila['distancia_km']}")

    if len(lista)== 0:
        raise ValueError
    else:
        return lista

def preprocesar(entrada: Entrada) -> list[float]:
    # TODO
    peso = entrada.peso_kg
    nuevo_peso = round(peso,1)
    distancia = entrada.distancia_km
    return [nuevo_peso, float(distancia)]


lectura = leer_csv('./data/raw/paquetes.csv')
print (lectura)
for elem in lectura:
    print(preprocesar(elem))
```
